```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def get_histogram(path, type_):
    df = pd.read_csv(f"{path}_{type_}.csv")
    plt.figure(figsize=(10, 4))
    df["f0"].hist(bins=100, density=True)
    #df["f0"].hist(bins=200)
    df["f0"].plot.kde()
    plt.title(f"Histogram - {type_}")
    plt.savefig(f"Histogram_{type_}.png")
    plt.show()

# Plot a kernel density estimate (KDE) curve
# df['value'].plot.density()
# # or
# df['value'].plot.kde()
```

```python
path = "../lists/GPS_cmd_pyin_setup_1"
get_histogram(path, "train")
#get_histogram(path, "val")
get_histogram(path, "test")
```

```python
def get_histogram(path):
    df_tr = pd.read_csv(f"{path}_train.csv")
    df_te = pd.read_csv(f"{path}_test.csv")
    plt.figure(figsize=(10, 4))
    df_tr["f0"].hist(bins=100, density=True)
    #df_tr["f0"].hist(bins=200)
    df_tr["f0"].plot.kde()
    df_te["f0"].hist(bins=100, density=True)
    df_te["f0"].plot.kde()
    plt.title(f"Histograms")
    plt.legend(["train", "test"])
    #plt.savefig(f"Histograms.png")
    plt.show()
```

```python
path = "../lists/GPS_cmd_pyin_setup_1"
get_histogram(path)
```
