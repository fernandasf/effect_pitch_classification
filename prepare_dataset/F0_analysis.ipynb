```python
import pandas as pd
import random
import numpy as np
import matplotlib.pyplot as plt
```

## Pyin

```python
path = "../pitch/GPS_cmd_16k_renamed_pyin_avg_by_word_user.csv"
df_raw = pd.read_csv(path)
df = df_raw.copy()
df["filepaths"] = df["filepaths"].apply(lambda x: x.replace("jovyan/work/OneDrive/Documentos/Doutorado/dataset", "ferreira/datasets"))
df
```

```python
labels = list(df['words'].unique())
for label in labels:
    df_label = df[df['words'] == label]
    plt.plot(df_label["f0"], label=f"{label}")
plt.legend()
plt.show()
```

```python
path = "../pitch/GPS_cmd_16k_renamed_pyin.csv"
df_raw = pd.read_csv(path)
df_o = df_raw.copy()
df_o["filepaths"] = df_o["filepaths"].apply(lambda x: x.replace("jovyan/work/OneDrive/Documentos/Doutorado/dataset", "ferreira/datasets"))
df_o
```

```python
labels = list(df_o['words'].unique())
for label in labels:
    df_label = df_o[df_o['words'] == label]
    plt.plot(df_label["f0_mean"], label=f"{label}")
plt.legend()
plt.show()
```

```python
label = "trabalho"
df_label = df_o[df_o['words'] == label]
df_label_F = df_label[df_label['genders'] == 'F']
df_label_M = df_label[df_label['genders'] == 'M']
plt.plot(df_label_M["f0_mean"], label=f"{label} - Male")
plt.plot(df_label_F["f0_mean"], label=f"{label} - Female")
plt.legend()
plt.show()
```
